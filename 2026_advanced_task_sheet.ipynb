## 2. Background

Quantitative trading strategies are often developed in stages:

1. A **baseline strategy** is implemented first to establish a reference point.
2. Strategy variations are then introduced to improve performance or manage risk.
3. All strategies are evaluated using consistent performance metrics.
4. Results are analysed to understand *why* one approach outperforms another.

In this assignment:

* You will implement an initial trading system based on a deterministic price-movement rule.
* You will then extend this system by introducing a **profit-taking mechanism**.
* Both systems will be tested on the **same historical price data** and time period to ensure a fair comparison.

This mirrors how real quantitative teams iterate on trading ideas before deployment.

### 4.2 Trading Algorithm 1 – Baseline Strategy (PR Hurdle 2)

You will implement **Trading System A**, which serves as the baseline trading strategy.

This strategy follows a deterministic price-movement rule and accumulates shares over time before exiting the position at the end of the trading period.

#### Trading Rules

* Initialise a variable `previous_price` to 0.
* On the **first trading day**:

  * If `previous_price` equals 0, execute a **buy** of 100 shares.
* On each subsequent trading day:

  * If the current closing price is **lower than the previous day’s closing price**, execute a **buy** of 100 shares.
* On the **final trading day**:

  * Sell **all accumulated shares** at the closing price.

### 4.3 Trading Algorithm 2 – Profit-Taking Strategy (PR Hurdle 3)

You will implement **Trading System B**, which extends Trading System A by introducing a profit-taking mechanism.

The purpose of this strategy is to reduce exposure when prices rise sufficiently, while still allowing the system to accumulate shares during price declines.

#### Trading Rules

In addition to the buying rules used in Trading System A, maintain the following internal variables:

* `accumulated_shares` — the total number of shares currently held
* `total_cost` — the total amount of capital spent acquiring the current holdings
* `average_purchase_price` — calculated as
  `total_cost / accumulated_shares` (when `accumulated_shares > 0`)

The profit-taking logic is defined as follows:

* On each trading day, if:

  * the current price is greater than
    `average_purchase_price × (1 + 20%)`,
* then:

  * sell **half of the current share holdings**,
  * reduce `total_cost` proportionally based on the average purchase price of the shares sold.

Additional rules:

* Buying behaviour must remain **identical** to Trading System A.
* On the **final trading day**, sell **any remaining shares**.
